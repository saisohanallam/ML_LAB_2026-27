```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.array([1, 8])
y = np.array([3, 10])

plt.plot(x, y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.figure(figsize=(8,5))
plt.plot(x,y,color='blue',linestyle='-.',marker='o',linewidth=3)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("line chart")
plt.legend(['my data series'])
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[5,2,3,4,6]
plt.figure(figsize=(8,5))
plt.bar(categories,values,color='#FF0000')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar chart')
plt.show()
```

```python
data = [12, 15, 13, 17, 16, 16, 19, 12, 14, 16, 14, 17, 18, 18, 15, 14]
plt.hist(data, bins=16, color='#008AFF')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Simple Histogram Example')
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 6, 8]
plt.scatter(x, y)
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Simple Scatter Plot Example')
plt.show()
```

```python
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Simple Pie Chart Example')
plt.axis('equal')
plt.show()
```

```python
data = [7, 8, 5, 6, 7, 9, 12, 5, 7, 8, 6]
plt.boxplot(data)
plt.title('Simple Box Plot Example')
plt.ylabel('Values')
plt.show()
```

```python
data = np.random.rand(10, 10)
plt.imshow(data, cmap='viridis', interpolation='nearest')
plt.colorbar()
plt.title('Simple Heatmap Example')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
%matplotlib inline
```

```python
from google.colab import files
uploaded = files.upload()  # opens a file picker, choose your dataset

filename = list(uploaded.keys())[0]
print(filename)
```

```python
# Use the correct reader for your file type
df = pd.read_csv(filename)        # for .csv
# df = pd.read_excel(filename)    # for .xlsx
# df = pd.read_json(filename)     # for .json

df.head()
```

```python
print(df.shape)
print(df.dtypes)
df.describe(include='all')
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
col = "money"  # replace with an actual column name
plt.figure(figsize=(8,5))
sns.histplot(df[col], kde=True)
plt.title(f"Distribution of {col}")
plt.show()
```

```python
col = "Time_of_Day"  # replace with an actual column name
plt.figure(figsize=(8,5))
df[col].value_counts().plot(kind="bar")
plt.title(f"Counts of {col}")
plt.xlabel(col)
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
x_col = "Date"
y_col = "Time"
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x=x_col, y=y_col)
plt.title(f"{y_col} vs {x_col}")
plt.show()
```
